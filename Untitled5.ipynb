```python
from google.colab import files
uploaded = files.upload()

# Dapatkan nama file yang diunggah
video_path = list(uploaded.keys())[0]
print(f"Video berhasil diunggah: {video_path}")
```

```python
import cv2
from ultralytics import YOLO

# 1. Load model YOLO
model = YOLO("yolov8n.pt")

# 2. Buka file video input
cap = cv2.VideoCapture(video_path)
assert cap.isOpened(), "Error saat membaca file video"

# 3. Ambil properti video (lebar, tinggi, FPS)
w, h, fps = (int(cap.get(x)) for x in (cv2.CAP_PROP_FRAME_WIDTH, cv2.CAP_PROP_FRAME_HEIGHT, cv2.CAP_PROP_FPS))

# 4. Tentukan posisi garis hitung (Horizontal di tengah layar)
line_y = h // 2

# 5. Atur Video Writer untuk menyimpan hasil video
output_video_path = "hasil_penghitungan.mp4"
video_writer = cv2.VideoWriter(output_video_path, cv2.VideoWriter_fourcc(*"mp4v"), fps, (w, h))

# Dictionary untuk menyimpan posisi Y sebelumnya berdasarkan ID objek
track_history = {}
total_count = 0

print("Sedang memproses video... Mohon tunggu.")
while cap.isOpened():
    success, frame = cap.read()
    if not success:
        print("Pemrosesan video selesai atau tidak ada frame tersisa.")
        break

    # Lakukan tracking dengan model YOLO
    results = model.track(frame, persist=True, show=False, verbose=False)

    # Gambar garis pembatas berwarna biru
    cv2.line(frame, (0, line_y), (w, line_y), (255, 0, 0), 3)

    if results[0].boxes.id is not None:
        boxes = results[0].boxes.xyxy.cpu().tolist()
        track_ids = results[0].boxes.id.int().cpu().tolist()
        clss = results[0].boxes.cls.int().cpu().tolist()

        for track_id, box, cls in zip(track_ids, boxes, clss):
            # Ambil titik pusat objek (x_center, y_center)
            x_center = int((box[0] + box[2]) / 2)
            y_center = int((box[1] + box[3]) / 2)

            # Gambar kotak pembatas dan label ID objek
            cv2.rectangle(frame, (int(box[0]), int(box[1])), (int(box[2]), int(box[3])), (0, 255, 0), 2)
            cv2.putText(frame, f"ID: {track_id}", (int(box[0]), int(box[1]) - 10), cv2.FONT_HERSHEY_SIMPLEX, 0.5, (0, 255, 0), 2)
            cv2.circle(frame, (x_center, y_center), 4, (0, 0, 255), -1)

            # Logika Penghitungan ketika melewati garis (Y_center berpindah memotong line_y)
            if track_id in track_history:
                prev_y = track_history[track_id]

                # Deteksi jika objek turun (melewati garis dari atas ke bawah)
                if prev_y < line_y <= y_center:
                    total_count += 1
                # Deteksi jika objek naik (melewati garis dari bawah ke atas)
                elif prev_y > line_y >= y_center:
                    total_count += 1

            # Update histori posisi Y objek
            track_history[track_id] = y_center

    # Tampilkan teks total hitungan pada layar video hasil
    cv2.putText(frame, f"Total Terhitung: {total_count}", (30, 70), cv2.FONT_HERSHEY_SIMPLEX, 1.5, (0, 0, 255), 3, cv2.LINE_AA)

    # Tulis frame ke video output
    video_writer.write(frame)

# 8. Bersihkan resource
cap.release()
video_writer.release()
cv2.destroyAllWindows()
print(f"Selesai! Video hasil disimpan sebagai: {output_video_path}")
```
